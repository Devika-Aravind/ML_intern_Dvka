## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import os
import sys
import subprocess
import pandas as pd

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/Devika-Aravind/ML_intern_Dvka"
REPO_DIR = "ML_intern_Dvka"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)

print("Working dir:", os.getcwd())
assert os.path.exists("data/raw/content_refresh_anonymized.csv"), "CSV not found — check repo cloned correctly"

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# 1. How many pages are stale AND still visible? (candidates for refresh)
stale_visible = ((df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)).sum()
print(f"Stale + visible pages (refresh candidates): {stale_visible} out of {len(df)}")

# 2. What percent of pages are declining?
declining_pct = (df["trend_direction"] == "down").mean() * 100
print(f"Percent of pages declining: {declining_pct:.1f}%")

# 3. Average impressions for declining vs growing pages
avg_impressions = df.groupby("trend_direction")["impressions_90d"].mean()
print("\nAverage impressions by trend direction:")
print(avg_impressions.round(0))

```

Out of 30,000 pages, only 17 are both stale and highly visible. A narrow,edge-case slice. But the the bigger signal is that 54.2% of all pages are currently declining, and these declining pages still average nearly 4,919 impressions each - meaning they still have real traffic worth protecting. This broader declining signal is what makes this lane worth pursuing for the next 7 weeks.
